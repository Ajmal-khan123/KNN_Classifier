### Measuring Success: Training and Testing Data

```python
X_train, X_test, y_train, y_test = train_test_split(
    iris_dataset['data'], iris_dataset['target'],
    random_state=0
)
```

```python
knn = KNeighborsClassifier(n_neighbors=5, algorithm='ball_tree', n_jobs=1, metric='minkowski', p=2, leaf_size=33, weights='uniform')
knn
```

```python
model_knn = knn.fit(X_train, y_train)
model_knn
```

```python
X_new = np.array([[5.9, 3. , 4.2, 1.5]])
print("X_new.shape: {}".format(X_new.shape))
```

```python
sns.scatterplot(X_new)
```

```python
prediction = knn.predict(X_new)
print("Prediction = .\n{}".format(prediction))
print("The predicted target name = .\n{}".format(iris_dataset['target_names'][prediction]))
```

## Evaluating the Model


```python
y_pred =knn.predict(X_test)
print('The predicted dataset for the data = .\n{}'.format(y_pred))
```

```python
print("The test score .\n{:.2f}".format(np.mean(y_pred == y_test)*100),"%")
```

```python
print("The score .\n{:.2f}".format(knn.score(X_test, y_test)*100), "%")
```
